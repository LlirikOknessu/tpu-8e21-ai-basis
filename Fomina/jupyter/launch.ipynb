```python
import yaml
import subprocess
```

```python
with open('dvc.yaml', 'r') as file:
    config = yaml.safe_load(file)
```

```python
for stage_name, stage in config['stages'].items():
    print(f'Выполнение стадии: {stage_name}')
    command = stage['cmd']

    process = subprocess.run(command, shell=True, capture_output=True, text=True)

    print('Вывод:', process.stdout)
    if process.stderr:
        print('Ошибка:', process.stderr)
```
